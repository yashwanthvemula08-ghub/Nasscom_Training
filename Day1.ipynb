Exception Handling

```python
# Raise your own error using 'raise'

def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    else:
        print("Age is valid.")

try:
    check_age(-5)
except ValueError as e:
    print("Caught an exception:", e)
```
